### Here we are considering  that if there is missing values that means the customer got no offer 

```python
df["Offer"] = df["Offer"].fillna(df["Offer"].mode()[0])
```

### Internet Type
### Missing values mean the customer does not have internet service
### Avg Monthly GB Download
### For customers with no internet service, data usage should be 0

```python

df["Internet Type"] = df["Internet Type"].fillna("No Internet")



df.loc[
    (df["Internet Type"] == "No Internet") & 
    (df["Avg Monthly GB Download"].isna()),
    "Avg Monthly GB Download"
] = 0
```

###  Multiple Lines
### Missing value means the customer does not have phone service
### Avg Monthly Long Distance Charges
### Customers without phone service cannot have long-distance charges

```python

df["Multiple Lines"] = df["Multiple Lines"].fillna("No Phone Service")



df.loc[
    (df["Multiple Lines"] == "No Phone Service") &
    (df["Avg Monthly Long Distance Charges"].isna()),
    "Avg Monthly Long Distance Charges"
] = 0
```

### If there is no internet service then this columns will have missing values so we replaced it with no internet serviece 

```python
internet_cols = [
    "Online Security",
    "Online Backup",
    "Device Protection Plan",
    "Premium Tech Support",
    "Streaming TV",
    "Streaming Movies",
    "Streaming Music",
    "Unlimited Data"
]

for col in internet_cols:
    df[col] = df[col].fillna("No Internet Service")
```

```python
df.drop(columns=["Churn Category", "Churn Reason"], inplace=True)
```

```python
# Save cleaned dataset as a new file
df.to_csv("Cleaned_Customer_Data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
